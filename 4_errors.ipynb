## Logical Errors

Logical errors are usually the most difficult errors to deal with as the code will still run without any errors and you do not get any hints on what is wrong. 
This PA includes a .py files with code that contains logical errors: `max_square_even.py`. Try using the Python debugger for these, as then you can see how the variables change step-by-step!

```python
# The code below is supposed to find the maximum even number in a list and return its square
numbers = [3, 7, 2, 8, 5, 10, 6]

def max_even_square(nums):
    max_num = 0
    for n in nums:
        if n % 2 == 1: # should by definition of an even number and the modulus % operator read: if n % 2 == 0 as % returns the rest after division
            if n > max_num:
                max_num = n
    square = max_num * 2 # the square operator in python is defined with double asterix **
    return square

print("The max even number squared is:", max_even_square(numbers))
```
